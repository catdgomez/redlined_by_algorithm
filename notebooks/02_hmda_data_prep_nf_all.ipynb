```python
hmda_all_melted = hmda_all_melted.drop(columns=["aus_values_0", "aus_types"])
```

```python
# group by the unique_id column, sum those values together, and reset the index
hmda_all_melted = hmda_all_melted.groupby("unique_id").sum().reset_index()
```

#### Create Hispanic vs. Not Hispanic 

```python
# applicant_ethnicity-1
# Description: Ethnicity of the applicant or borrower
# Values:
# 1 - Hispanic or Latino
# 11 - Mexican
# 12 - Puerto Rican
# 13 - Cuban
# 14 - Other Hispanic or Latino
# 2 - Not Hispanic or Latino
# 3 - Information not provided by applicant in mail, internet, or telephone application
# 4 - Not applicable
```

#### Exclude info not provided and not applicable

```python
hmda_all = hmda_all[~hmda_all['applicant_ethnicity-1'].isin([3, 4])]
```

```python
# ethnicity columns
ethnicity_cols = ['applicant_ethnicity-1', 'applicant_ethnicity-2', 'applicant_ethnicity-3',
                  'applicant_ethnicity-4', 'applicant_ethnicity-5']
hispanic_nums = [1, 11, 12, 13, 14]

# indicator columns for each hispanic subcategory --> one per num
for num in hispanic_nums:
    hmda_all[f'ae_values_{num}'] = (hmda_all[ethnicity_cols] == num).any(axis=1)

# overall hispanic indicator -- 1 = Hispanic, 0 = Not Hispanic
hmda_all['is_hispanic'] = hmda_all[[f'ae_values_{num}' for num in hispanic_nums]].any(axis=1).astype(int)

# not hispanic indicator -- 1 = Not Hispanic, 0 = Hispanic
hmda_all['not_hispanic'] = (~hmda_all['is_hispanic'].astype(bool)).astype(int)

# hispanic vs. not hispanic label column
hmda_all['ethnicity_group'] = np.where(hmda_all['is_hispanic'], 'Hispanic', 'Not Hispanic')
```
